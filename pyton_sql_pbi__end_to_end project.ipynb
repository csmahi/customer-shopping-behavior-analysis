#fixing nulls

```python
df['Review Rating']= df.groupby('Category')['Review Rating'].transform(lambda x:x.fillna(x.median()))
```

```python
df['Review Rating'].isna().sum()
```

```python
df.columns
```

```python
df.columns = df.columns.str.lower()
```

```python
df.columns= df.columns.str.replace(' ','_')
```

```python
df= df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
##create column age group

labels = ['young adult','adult','middleaged','senior']
df['age_group'] = pd.qcut(df['age'],4,labels=labels)
```

```python
df = df.drop('promo_code_used', axis=1)
```

```python
df['frequency_of_purchases'].head(20)
```

```python
## create a column for purchase frequency days
frequency_mapping ={
'Fortnightly': 14,
 'Weekly': 7,
 'Bi-Weekly':14,
 'Monthly':30,
 'Quarterly': 90,
 'Annually':365,
 'Every 3 Months':90}
 
df['purchase_frequency_days'] =  df['frequency_of_purchases'].map(frequency_mapping)
```
